```python

# Step 1 — Project setup and load datasets

import os
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a configurable dataset path
DATA_DIR = Path(
    os.environ.get(
        "OLIST_DATA_DIR",
        "/content/drive/MyDrive/Ecommerce_Project/olist_cleaned_data_updated"
    )
)

# Check that all required files exist
required_files = [
    "olist_orders_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_customers_dataset.csv",
    "olist_sellers_dataset.csv",
]

missing_files = [
    filename for filename in required_files
    if not (DATA_DIR / filename).is_file()
]

if missing_files:
    raise FileNotFoundError(
        f"Missing files in {DATA_DIR}:\n{missing_files}\n"
        "Please check your Google Drive folder path."
    )

# Load datasets
orders = pd.read_csv(DATA_DIR / "olist_orders_dataset.csv")
order_items = pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv")
customers = pd.read_csv(DATA_DIR / "olist_customers_dataset.csv")
sellers = pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv")

print("Dataset folder:", DATA_DIR)
print("Orders:", orders.shape)
print("Order items:", order_items.shape)
print("Customers:", customers.shape)
print("Sellers:", sellers.shape)
```

### Step 2 — Initial Data Exploration and Inspection
Let's take a look at the structures, missing values, and data types of each loaded dataset.

```python
for name, df in [('Orders', orders), ('Order Items', order_items), ('Customers', customers), ('Sellers', sellers)]:
    print(f"=== {name} Dataset ===")
    print(df.info())
    print(f"Missing values:\n{df.isnull().sum()}\n")
```

### Step 3 — Merge Datasets
Now we will merge the `orders`, `order_items`, `customers`, and `sellers` datasets into a unified dataframe to facilitate broader analytics.

```python
# Merge order items with orders first
merged_df = pd.merge(order_items, orders, on='order_id', how='inner')

# Merge with customer details
merged_df = pd.merge(merged_df, customers, on='customer_id', how='inner')

# Merge with seller details
merged_df = pd.merge(merged_df, sellers, on='seller_id', how='inner')

print("Merged Dataset shape:", merged_df.shape)
display(merged_df.head())
```
